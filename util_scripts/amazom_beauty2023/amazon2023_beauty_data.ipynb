```python
import pandas as pd
import numpy as np

from google.colab import drive

# Should be run on google colab with account that contains the data
drive.mount('/content/drive')

TRAIN_PATH = '/content/drive/MyDrive/RecSys data/All_Beauty.train.csv'
VALID_PATH = '/content/drive/MyDrive/RecSys data/All_Beauty.valid.csv'
```

```python
train_df = pd.read_csv(TRAIN_PATH)
valid_df = pd.read_csv(VALID_PATH)
```

```python
# convert ids to integers to save memory
all_user_ids = pd.concat([train_df['user_id'], valid_df['user_id']]).unique()
all_parent_asins = pd.concat([train_df['parent_asin'], valid_df['parent_asin']]).unique()

# Create mappings
user_id_mapping = {id_str: i+1 for i, id_str in enumerate(all_user_ids)}
parent_asin_mapping = {asin_str: i+1 for i, asin_str in enumerate(all_parent_asins)}

# Apply mappings to train_df
train_df['user_id'] = train_df['user_id'].map(user_id_mapping).astype(np.int32)
train_df['parent_asin'] = train_df['parent_asin'].map(parent_asin_mapping).astype(np.int32)

# Apply mappings to valid_df
valid_df['user_id'] = valid_df['user_id'].map(user_id_mapping).astype(np.int32)
valid_df['parent_asin'] = valid_df['parent_asin'].map(parent_asin_mapping).astype(np.int32)
```

```python
formatted_train_df = train_df[['user_id', 'parent_asin']].copy()
formatted_train_df['constant'] = 1

formatted_train_df = formatted_train_df.sort_values(by=['user_id', 'parent_asin']).reset_index(drop=True)

formatted_train_df.to_csv('train.txt', sep=' ', index=False, header=False)
```

```python
formatted_valid_df = valid_df[['user_id', 'parent_asin']].copy()
formatted_valid_df['constant'] = 1

formatted_valid_df = formatted_valid_df.sort_values(by=['user_id', 'parent_asin']).reset_index(drop=True)


formatted_valid_df.to_csv('test.txt', sep=' ', index=False, header=False)
```
